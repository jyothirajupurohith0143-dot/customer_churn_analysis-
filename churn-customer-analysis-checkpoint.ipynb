```python
Q1 = df['MonthlyCharges'].quantile(0.25)
Q3 = df['MonthlyCharges'].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)
```

```python
outliers = df[
    (df['MonthlyCharges'] < lower_limit) |
    (df['MonthlyCharges'] > upper_limit)
]

print("Number of outliers:", len(outliers))
```

```python
outliers[['customerID', 'MonthlyCharges']].head(20)
```

```python
from scipy.stats import zscore

z_scores = zscore(df['MonthlyCharges'])

outliers_z = df[abs(z_scores) > 3]

print("Number of Z-score outliers:", len(outliers_z))
```

```python
outliers_z[['customerID', 'MonthlyCharges']].head(20)
```
